```python
import pandas as pd
import tsfel
import numpy as np
```

```python
#label csv
df_label = pd.read_csv('../../name_label_kisa_new.csv')
```

```python
cfg = tsfel.get_features_by_domain()
#statictical 
#temporal 
#frequency
```

```python
df = pd.read_csv('../../new_posture_data_kisa_last.csv')
```

```python
unique_persons = df.iloc[:, 0].unique()
```

```python
all_features = []
all_labels = []
all_names = []

for person in unique_persons:

    person_data = df[df.iloc[:, 0] == person]

    # Neck ve Torso serileri
    neck_series = person_data["Neck_Inclination_2D"].reset_index(drop=True)
    torso_series = person_data["Torso_Inclination_2D"].reset_index(drop=True)
    neck_series_3d = person_data["Neck_Inclination_3D"].reset_index(drop=True)
    torso_series_3d = person_data["Torso_Inclination_3D"].reset_index(drop=True)
    shoulder_height_diff_series = person_data["Shoulder_Height_Diff"].reset_index(drop=True)
    head_tilt_2d_series = person_data["Head_Tilt_2D"].reset_index(drop=True)
    head_tilt_3d_series = person_data["Head_Tilt_3D"].reset_index(drop=True)
    shoulder_line_angle_2d_series = person_data["Shoulder_Line_Angle_2D"].reset_index(drop=True)
    shoulder_line_angle_3d_series = person_data["Shoulder_Line_Angle_3D"].reset_index(drop=True)


    #feature extraction
    features_neck = tsfel.time_series_features_extractor(cfg, neck_series.to_frame(), verbose=0)
    features_torso = tsfel.time_series_features_extractor(cfg, torso_series.to_frame(), verbose=0)
    features_neck_3d = tsfel.time_series_features_extractor(cfg, neck_series_3d.to_frame(), verbose=0)
    features_torso_3d = tsfel.time_series_features_extractor(cfg, torso_series_3d.to_frame(), verbose=0)
    features_shoulder_height_diff = tsfel.time_series_features_extractor(cfg, shoulder_height_diff_series.to_frame(), verbose=0)
    features_head_tilt_2d = tsfel.time_series_features_extractor(cfg, head_tilt_2d_series.to_frame(), verbose=0)
    features_head_tilt_3d = tsfel.time_series_features_extractor(cfg, head_tilt_3d_series.to_frame(), verbose=0)
    features_shoulder_line_angle_2d = tsfel.time_series_features_extractor(cfg, shoulder_line_angle_2d_series.to_frame(), verbose=0)
    features_shoulder_line_angle_3d = tsfel.time_series_features_extractor(cfg, shoulder_line_angle_3d_series.to_frame(), verbose=0)

    combined_features = pd.concat([features_neck, features_torso, features_neck_3d, features_torso_3d, features_shoulder_height_diff, features_head_tilt_2d, features_head_tilt_3d, features_shoulder_line_angle_2d, features_shoulder_line_angle_3d ], axis=1)

    all_features.append(combined_features.iloc[0].values)

    #label
    label = df_label[df_label['Video'] == person + '.mp4']['Label'].iloc[0]
    all_labels.append(label)

    #video
    video = df_label[df_label['Video'] == person + '.mp4']['Video'].iloc[0]
    all_names.append(video)

features_array = np.array(all_features)
print(features_array.shape)
features_df = pd.DataFrame(features_array, columns=combined_columns)
features_df['label'] = all_labels
features_df['video'] = all_names

#save
features_df.to_csv('tsfel_features_short_new.csv', index=False)

```
